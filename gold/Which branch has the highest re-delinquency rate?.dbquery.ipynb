```sql
SELECT branch_id, branch_name, region,
       SUM(total_events)          AS events,
       SUM(total_redelinquencies) AS redelinquencies,
       ROUND(SUM(total_redelinquencies) * 100.0 / SUM(total_events), 2) AS rate_pct
FROM banking.gold.agg_redelinquency_branch_product
GROUP BY branch_id, branch_name, region
ORDER BY rate_pct DESC, redelinquencies DESC
LIMIT 5;
-- top row: BR004 Jeddah Tahlia, Makkah region, 3 events, 2 re-delinquencies, 66.67%
```
