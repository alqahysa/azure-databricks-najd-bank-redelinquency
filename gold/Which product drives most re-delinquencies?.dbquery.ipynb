```sql
SELECT product_type, product_code, product_desc,
       SUM(total_redelinquencies) AS redelinquencies,
       ROUND(SUM(total_redelinquencies) * 100.0
             / SUM(SUM(total_redelinquencies)) OVER (), 2) AS share_of_all_pct
FROM banking.gold.agg_redelinquency_branch_product
GROUP BY product_type, product_code, product_desc
HAVING SUM(total_redelinquencies) > 0
ORDER BY redelinquencies DESC, product_code;
-- PF02 and CC02 lead with 2 each (25% each); by product_type, Credit Card = 4 of 8 (50%)
```
