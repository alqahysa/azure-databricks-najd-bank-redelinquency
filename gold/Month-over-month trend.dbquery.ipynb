```sql
WITH m AS (
  SELECT due_month,
         COUNT(*)                    AS events,
         SUM(redelinquency_30d_flag) AS redelinquencies
  FROM banking.gold.fact_redelinquency_event
  GROUP BY due_month
)
SELECT due_month, events, redelinquencies,
       ROUND(redelinquencies * 100.0 / events, 2)          AS rate_pct,
       redelinquencies - LAG(redelinquencies) OVER (ORDER BY due_month) AS mom_change
FROM m
ORDER BY due_month;
-- 2026-06: 6 / 0 | 2026-07: 8 / 3 | 2026-08: 7 / 3 | 2026-09: 7 / 2 | 2026-10: 2 / 0
```
